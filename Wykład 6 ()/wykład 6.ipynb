## Tworzenie napisów

Napisy możemy zapisywać w pojedynczych lub podwójnych cudzysłowach:

```python
name = "Python"
university = 'AGH'
```

Obie formy tworzą obiekt typu `str`.

````python
name = "Python"
university = 'AGH'

print(name, type(name))
print(university, type(university))
````

## Dobieraj cudzysłów do tekstu

Możemy wykorzystać drugi rodzaj cudzysłowu wewnątrz napisu:

```python
message = "Python's syntax"
quote = 'Powiedział: "Witaj"'
```

Dzięki temu kod jest czytelniejszy.

````python
message = "Python's syntax"
quote = 'Powiedział: "Witaj"'

print(message)
print(quote)
````

## Znaki specjalne

W napisach możemy używać sekwencji specjalnych:

| Zapis | Znaczenie |
|---|---|
| `\n` | nowa linia |
| `\t` | tabulator |
| `\\` | znak `\` |
| `\"` | cudzysłów |

Znak `\` rozpoczyna sekwencję specjalną.

````python
text = "Pierwsza linia\nDruga linia\n\tWcięcie"

print(text)
````

## Napisy wieloliniowe

Potrójny cudzysłów pozwala zapisać napis obejmujący wiele linii:

```python

```

To nadal jest zwykły obiekt `str`.

W odpowiednim miejscu taki napis może pełnić rolę docstringa, co widzieliśmy przy funkcjach.

````python
message = """Pierwsza linia
Druga linia
Trzecia linia"""

print(message)
print(type(message))
````

## Raw string — gdy `\` ma być zwykłym znakiem

Prefiks `r` ogranicza interpretację sekwencji specjalnych:

```python
path = r"C:\Users\student\data"
```

Przydaje się m.in. przy ścieżkach i później przy wyrażeniach regularnych.

````python
normal = "C:\\Users\\student\\data"
raw = r"C:\Users\student\data"

print(normal)
print(raw)
````

## `str` jest sekwencją

Podobnie jak lista i krotka, napis posiada uporządkowane elementy.

Dla:

```python
text = "Python"
```

możemy myśleć:

```text
 P  y  t  h  o  n
 0  1  2  3  4  5
-6 -5 -4 -3 -2 -1
```

````python
text = "Python"

print(text[0])
print(text[1])
print(text[-1])
print(len(text))
````

## Slicing napisów

Składnia jest taka sama jak dla list:

```python
text[start:stop]
text[start:stop:step]
```

Element `stop` nie należy do wyniku.

````python
text = "Programowanie"

print(text[0:7])
print(text[:7])
print(text[7:])
print(text[::2])
print(text[::-1])
````

## Napis jest niemutowalny

Po utworzeniu obiektu `str` nie możemy zmienić pojedynczego znaku:

```python
text = "Python"
text[0] = "J"      # TypeError
```

To ta sama idea niemutowalności, którą poznaliśmy przy krotkach.

````python
text = "Python"

# Odkomentuj, aby zobaczyć TypeError:
# text[0] = "J"

print(text)
````

## Jak więc „zmienić” napis?

Tworzymy **nowy obiekt**:

```python
text = "Python"
text = "J" + text[1:]
```

Nazwa `text` zostaje związana z nowym napisem.

````python
text = "Python"
print(text)

text = "J" + text[1:]
print(text)
````

## Operatory na napisach

Napisy obsługują m.in.:

```python
+     # łączenie
*     # powtarzanie
in    # sprawdzenie zawartości
```

````python
first = "AGH"
second = "Kraków"

print(first + " " + second)
print("ha" * 3)
print("Krak" in second)
print("Warszawa" not in second)
````

## Metody nie modyfikują istniejącego napisu

Ponieważ `str` jest niemutowalny:

```python
text.upper()
```

zwraca **nowy napis**.

Oryginalny obiekt się nie zmienia.

````python
text = "Python"

upper_text = text.upper()

print("oryginał:", text)
print("wynik:", upper_text)
````

## Wielkość liter

Najczęściej używane metody:

```python
lower()
upper()
capitalize()
title()
casefold()
```

`casefold()` służy głównie do bardziej niezawodnego porównywania tekstu bez rozróżniania wielkości liter.

````python
text = "pYtHoN PROGRAMMING"

print(text.lower())
print(text.upper())
print(text.capitalize())
print(text.title())
print(text.casefold())
````

## Porównanie tekstu bez wielkości liter

Zamiast:

```python
answer.lower() == "tak"
```

dla tekstu użytkownika możemy użyć:

```python
answer.casefold() == "tak"
```

````python
answer = "TAK"

if answer.casefold() == "tak":
    print("Użytkownik odpowiedział tak")
````

## Białe znaki

Dane tekstowe bardzo często zawierają niepotrzebne spacje lub znaki końca linii.

Przydatne są:

```python
strip()
lstrip()
rstrip()
```

````python
text = "   temperatura = 21.7   \n"

print(repr(text))
print(repr(text.strip()))
print(repr(text.lstrip()))
print(repr(text.rstrip()))
````

## Dlaczego używamy `repr()`?

`print()` interpretuje znaki specjalne.

`repr()` pokazuje reprezentację obiektu, dzięki czemu widzimy np.:

```text

\t
spacje
```

To bardzo przydatne przy diagnozowaniu problemów z tekstem.

````python
text = "A\tB\nC"

print(text)
print(repr(text))
````

## Wyszukiwanie w napisie

Najprostsze pytanie często zapisujemy przez `in`:

```python
"alarm" in message
```

Jeżeli potrzebujemy pozycji, możemy użyć `find()` lub `index()`.

````python
message = "sensor_01: alarm temperatury"

print("alarm" in message)
print(message.find("alarm"))
print(message.find("awaria"))
````

## `find()` a `index()`

```python

```

zwraca `-1`.

Natomiast:

```python
text.index("xyz")
```

zgłasza `ValueError`.

Jeżeli pytamy tylko „czy występuje?”, najczytelniejsze jest zwykle `in`.

## Walidacja prostych napisów

Przydatne metody:

```python

isalnum()

```

Zwracają `True` lub `False`.

```python

```

daje `True`, ale:

```python

```

Metody `is...` nie zastępują pełnej walidacji danych.

````python
print("123".isdigit())
print("21.7".isdigit())
print("-10".isdigit())
````

## Dzielenie tekstu — `split()`

Jedna z najważniejszych metod przy przetwarzaniu danych tekstowych:

```python
text.split(separator)
```

zwraca listę fragmentów.

````python
line = "sensor_01;temperature;21.7;C"

parts = line.split(";")

print(parts)
print(type(parts))
````

## Przetwarzanie tekstowego rekordu

Teraz możemy nadać fragmentom znaczenie.

````python
line = "sensor_01;temperature;21.7;C"

parts = line.split(";")

sensor = parts[0]
measurement = parts[1]
value = float(parts[2])
unit = parts[3]

print(sensor)
print(measurement)
print(value, type(value))
print(unit)
````

## Rozpakowanie jest czytelniejsze

Ponieważ `split()` zwraca listę, możemy wykorzystać rozpakowanie:

```python
sensor, measurement, value, unit = line.split(";")
```

Następnie wykonujemy konwersję wartości.

````python
line = "sensor_01;temperature;21.7;C"

sensor, measurement, value, unit = line.split(";")
value = float(value)

print(sensor, measurement, value, unit)
````

## Łączenie tekstu — `join()`

Operacją odwrotną do `split()` jest często `join()`.

```python
separator.join(collection)
```

Przykład:

````python
words = ["Python", "jest", "czytelny"]

sentence = " ".join(words)

print(sentence)
````

## Dlaczego składnia `" ".join(words)` wygląda dziwnie?

Obiektem wykonującym operację jest separator:

```python
" ".join(words)
";".join(values)
",".join(columns)
```

Czyli:

> „połącz elementy, używając mnie jako separatora”.

````python
values = ["sensor_01", "temperature", "21.7", "C"]

print(";".join(values))
print(",".join(values))
````

## `split()` + przetwarzanie + `join()`

To bardzo częsty schemat:

```text
tekst wejściowy
↓
split()
↓
lista elementów
↓
przetwarzanie
↓
join()
↓
tekst wynikowy
```

````python
line = "  Python   jest   czytelny  "

words = line.split()
clean_text = " ".join(words)

print(repr(line))
print(repr(clean_text))
````

## Formatowanie tekstu — f-string

Najczęściej używany obecnie sposób tworzenia komunikatów:

```python
name = "sensor_01"
value = 21.7

message = f"{name}: {value} °C"
```

````python
sensor = "sensor_01"
value = 21.7

message = f"{sensor}: {value} °C"

print(message)
````

## Formatowanie liczb

F-string może określać sposób prezentacji liczby:

```python
f"{value:.2f}"
```

oznacza dwie cyfry po separatorze dziesiętnym.

````python
value = 21.73642

print(f"{value}")
print(f"{value:.1f}")
print(f"{value:.2f}")
print(f"{value:8.2f}")
````

## Formatowanie procentów i szerokości

Kilka praktycznych przykładów:

````python
ratio = 0.8734
name = "sensor"

print(f"{ratio:.1%}")
print(f"|{name:<12}|")
print(f"|{name:^12}|")
print(f"|{name:>12}|")
````

## f-string jako narzędzie diagnostyczne

Zapis:

```python
f"{value=}"
```

wypisuje nazwę wyrażenia i jego wartość.

To wygodne podczas prostego debugowania.

````python
temperature = 21.7
limit = 30

print(f"{temperature=}")
print(f"{limit=}")
````

## Kodowanie tekstu — `encode()`

Aby zamienić tekst na bajty, potrzebujemy **kodowania**.

Najczęściej używamy UTF-8:

```python
data = text.encode("utf-8")
```

````python
text = "zażółć"

data = text.encode("utf-8")

print(text)
print(data)

print(type(text))
print(type(data))
````

## Dekodowanie — `decode()`

Operacja odwrotna:

```python
text = data.decode("utf-8")
```

````python
data = b"Python"

text = data.decode("utf-8")

print(data, type(data))
print(text, type(text))
````

## Znak i bajty to nie to samo

Porównajmy długość napisu i jego reprezentacji UTF-8:

````python
text = "ą"

data = text.encode("utf-8")

print(text, len(text))
print(data, len(data))
````

## Dlaczego kodowanie ma znaczenie?

Ten sam tekst może być reprezentowany jako różne ciągi bajtów zależnie od kodowania.

Jeżeli program zapisuje dane w jednym kodowaniu, a odczytuje w innym, pojawiają się błędy lub nieprawidłowe znaki.

W praktyce UTF-8 jest dzisiaj najczęstszym wyborem.

Do kodowania wrócimy przy pracy z plikami.

## Niemutowalność i wydajność

Rozważ budowanie dużego napisu przez wielokrotne:

```python
result += word
```

Ponieważ `str` jest niemutowalny, kolejne operacje mogą wymagać tworzenia nowych obiektów.

Dla wielu fragmentów często lepszym narzędziem jest `join()`.

````python
%%timeit
words = ["Python"] * 1000
result = ""
for word in words:
    result += word
````

````python
%%timeit
result = "".join(words)
````


## Przykład integrujący

Otrzymujemy tekst:

```text
sensor_01;temperature;21.736;C
```

Chcemy wygenerować czytelny komunikat:

```text
sensor_01: temperature = 21.7 °C
```

````python
line = "sensor_01;temperature;21.736;C"

sensor, measurement, value, unit = line.strip().split(";")
value = float(value)

message = f"{sensor}: {measurement} = {value:.1f} °{unit}"

print(message)
````

## Przetwarzanie wielu rekordów

Łączymy wiedzę z poprzednich wykładów:

- listy,
- pętle,
- funkcje,
- napisy,
- konwersję typów.

````python
def parse_measurement(line):
    sensor, measurement, value, unit = line.strip().split(";")

    return {
        "sensor": sensor,
        "measurement": measurement,
        "value": float(value),
        "unit": unit,
    }


lines = [
    "sensor_01;temperature;21.7;C",
    "sensor_02;temperature;22.3;C",
    "sensor_03;humidity;45.2;%",
]

for line in lines:
    measurement = parse_measurement(line)
    print(measurement)
````

## Funkcja formatująca wynik

Możemy rozdzielić odpowiedzialności:

- jedna funkcja **parsuje dane**,
- druga **formatuje komunikat**.

````python
def format_measurement(data):
    return (
        f'{data["sensor"]}: '
        f'{data["measurement"]} = '
        f'{data["value"]:.1f} {data["unit"]}'
    )


line = "sensor_01;temperature;21.736;C"

data = parse_measurement(line)

print(format_measurement(data))
````
