## `print()` a `return`

To jedna z najważniejszych różnic w tym wykładzie.

`print()`:

> pokazuje wartość użytkownikowi.

`return`:

> zwraca wartość do miejsca, z którego funkcja została wywołana.

````python
def add_and_print(a, b):
    print(a + b)

result = add_and_print(2, 3)

print("Zmienna result:", result)
````

## Funkcja bez `return` zwraca `None`

Poprzednia funkcja wypisała `5`, ale nie zwróciła wyniku.

Dlatego:

```python
result = add_and_print(2, 3)
```

wiąże nazwę `result` z wartością `None`.

Jeżeli wynik ma być dalej używany przez program, powinniśmy go **zwrócić**.

````python
def add(a, b):
    return a + b

result = add(2, 3)

print("Wynik:", result)
print("Podwojony wynik:", result * 2)
````

## `return` kończy wykonanie funkcji

Po wykonaniu `return` Python opuszcza funkcję.

````python
def classify_temperature(temperature):
    if temperature < -273.15:
        return "błędny pomiar"

    if temperature > 30:
        return "wysoka"

    return "normalna"

print(classify_temperature(-300))
print(classify_temperature(35))
print(classify_temperature(20))
````

## Funkcje pomagają rozkładać problem

Dobrym pytaniem podczas pisania programu jest:

> Jakie mniejsze zadania składają się na mój problem?

Przykład dla pomiarów temperatury:

1. sprawdź poprawność pomiaru,
2. sprawdź, czy przekroczono limit,
3. przeanalizuj serię pomiarów.

Każde z tych zadań może stać się osobną funkcją.

````python
def is_valid_temperature(temperature):
    return temperature >= -273.15


def is_alarm(temperature, limit):
    return temperature > limit
````

````python
measurements = [21, 31, -300, 24, 36]
limit = 30

for value in measurements:
    if not is_valid_temperature(value):
        continue

    if is_alarm(value, limit):
        print("ALARM:", value)
````

## Zakres nazw

Nazwa utworzona wewnątrz funkcji jest zazwyczaj **lokalna** dla tej funkcji.

````python
temperature = 20

def example():
    result = 10
    print("W funkcji:", temperature)
    print("W funkcji:", result)

example()

print("Poza funkcją:", temperature)

# Odkomentuj, aby zobaczyć NameError:
# print(result)
````

## Nazwy lokalne i nazwy modułu

Uproszczony model:

```text
poziom modułu
temperature

└── funkcja example()
    └── result
```

Funkcja może odczytać nazwę zdefiniowaną na poziomie modułu, ale nazwy utworzone wewnątrz funkcji nie są automatycznie dostępne poza nią.

## Zmienna globalna — można, ale ostrożnie

Python pozwala modyfikować nazwę z poziomu modułu za pomocą `global`.

````python
counter = 0

def increment_global():
    global counter
    counter += 1

increment_global()
print(counter)
````

## Często lepiej jawnie przekazać i zwrócić wartość

Zamiast ukrytej modyfikacji stanu:

```python
global counter
```

możemy zapisać:

```python
def increment(counter):
    return counter + 1

counter = increment(counter)
```

Taka funkcja ma mniej ukrytych zależności i zwykle łatwiej ją testować i rozumieć.

````python
def increment(counter):
    return counter + 1

counter = 0
counter = increment(counter)

print(counter)
````

## Funkcja i model obiektów z Wykładu 2

Pamiętamy:

```text
nazwa → obiekt
```

Podczas wywołania funkcji parametr zostaje związany z obiektem przekazanym jako argument.

To ma znaczenie szczególnie dla obiektów **mutowalnych**.

````python
def add_measurement(data, value):
    data.append(value)

measurements = [20, 21]

add_measurement(measurements, 22)

print(measurements)
````

## Dlaczego lista się zmieniła?

Uproszczony model:

```text
measurements ──┐
               ├──► [20, 21]
data ──────────┘
```

Wewnątrz funkcji parametr `data` wskazuje na ten sam obiekt listy.

Operacja:

```python
data.append(22)
```

modyfikuje ten obiekt.

## Obiekt niemutowalny zachowuje się inaczej

Liczby są niemutowalne.

````python
def increase(value):
    value += 1
    return value

number = 10
result = increase(number)

print("number:", number)
print("result:", result)
````

## Jak o tym myśleć?

Nie musimy mówić, że Python przekazuje argumenty „przez wartość” albo „przez referencję”.

W tym kursie używamy spójnego modelu:

> parametr funkcji zostaje związany z obiektem przekazanym jako argument.

Dalsze zachowanie zależy między innymi od tego, czy dany obiekt jest mutowalny.

## Gdy nie chcemy modyfikować przekazanej listy

Możemy utworzyć jej kopię.

````python
def add_measurement_copy(data, value):
    result = data.copy()
    result.append(value)
    return result

measurements = [20, 21]
new_measurements = add_measurement_copy(measurements, 22)

print("oryginał:", measurements)
print("wynik:", new_measurements)
````

## Ważna pułapka: mutowalny argument domyślny

Ten kod wygląda niewinnie:

```python
def add_value(value, values=[]):
    values.append(value)
    return values
```

Jakich wyników się spodziewasz po trzech wywołaniach?

````python
def add_value_bad(value, values=[]):
    values.append(value)
    return values

print(add_value_bad(1))
print(add_value_bad(2))
print(add_value_bad(3))
````

## Dlaczego wartości się „pamiętają”?

Argument domyślny jest tworzony podczas **definiowania funkcji**, a nie przy każdym jej wywołaniu.

W kolejnych wywołaniach używana jest więc ta sama lista.

To kolejny przykład, że:

> znajomość składni to dopiero pierwszy poziom rozumienia języka.

## Poprawny wzorzec z `None`

Zamiast mutowalnego obiektu jako wartości domyślnej używamy często `None`.

````python
def add_value(value, values=None):
    if values is None:
        values = []

    values.append(value)
    return values

print(add_value(1))
print(add_value(2))
print(add_value(3))
````

## Funkcja może zwracać więcej niż jedną wartość

Python pozwala wygodnie zwrócić kilka wartości naraz.

````python
def min_max(values):
    return min(values), max(values)

minimum, maximum = min_max([4, 1, 8, 3])

print(minimum)
print(maximum)
````

## Co naprawdę zwraca funkcja?

Zapis:

```python
return min(values), max(values)
```

zwraca **krotkę**.

Możemy więc również napisać:

```python
result = min_max([4, 1, 8, 3])
print(result)
print(type(result))
```

To łączy funkcje z krotkami poznanymi na Wykładzie 2.

````python
result = min_max([4, 1, 8, 3])

print(result)
print(type(result))
````

## Od programu liniowego do programu z funkcjami

Po Wykładzie 3 potrafiliśmy:

```python
for temperature in temperatures:
    if temperature > 30:
        print("ALARM")
```

Teraz możemy rozdzielić odpowiedzialności.

````python
def is_valid_temperature(temperature: float) -> bool:
    """Return True when temperature is physically valid."""
    return temperature >= -273.15


def is_alarm(temperature: float, limit: float = 30.0) -> bool:
    """Return True when temperature exceeds the limit."""
    return temperature > limit


def analyze_measurements(measurements, limit=30.0):
    """Print alarms for valid measurements exceeding the limit."""
    for index, temperature in enumerate(measurements, start=1):
        if not is_valid_temperature(temperature):
            continue

        if is_alarm(temperature, limit):
            print(f"Pomiar {index}: ALARM ({temperature} °C)")
````

````python
temperatures = [21, 31, -300, 24, 36]

analyze_measurements(temperatures)
````

## Co zyskaliśmy?

```text
↓

↓

analyze_measurements()
↓

```

Każdą funkcję można:

- zrozumieć osobno,
- przetestować osobno,
- wykorzystać ponownie,
- zmienić bez przepisywania całego programu.

## Podsumowanie

Po tym wykładzie powinieneś pamiętać:

- funkcje pozwalają dzielić program na mniejsze zadania,
- `def` definiuje funkcję, a zapis z `()` ją wywołuje,
- parametr występuje w definicji, argument w wywołaniu,
- `print()` wyświetla, a `return` zwraca wartość,
- `return` kończy wykonanie funkcji,
