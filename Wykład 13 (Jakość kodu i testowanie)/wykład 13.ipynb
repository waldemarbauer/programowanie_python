## Program może być „działający” i nadal zły

Kod może:

- działać tylko dla jednego przykładu,
- być trudny do przeczytania,
- zawierać ukryte zależności,
- psuć się po drobnej zmianie,
- być niemożliwy do przetestowania,
- dawać niepoprawne wyniki bez wyjątku.

Dlatego potrzebujemy pojęcia **jakości kodu**.

## Czytelność jest cechą techniczną

Kod czyta człowiek znacznie częściej niż go pisze.

Czytelny kod:

- ułatwia znalezienie błędu,
- ułatwia testowanie,
- zmniejsza ryzyko przypadkowej zmiany zachowania,
- ułatwia pracę zespołową.

## PEP 8

PEP 8 opisuje podstawowe konwencje stylu kodu Pythona.

Przykładowe zasady:

```text
snake_case      → funkcje i zmienne
PascalCase      → klasy
UPPER_CASE      → stałe
4 spacje        → wcięcie
czytelne odstępy i długość linii
```

Nie chodzi o estetykę samą w sobie — chodzi o wspólny język zespołu.

## Małe funkcje są łatwiejsze do testowania

Funkcja:

```python
def celsius_to_kelvin(value):
    return value + 273.15
```

ma:

- jedno wejście,
- jeden wynik,
- brak ukrytego stanu,
- łatwe przypadki testowe.

Takie funkcje są bardzo wdzięczne w testach.

## Efekty uboczne

Funkcja ma efekt uboczny, jeśli zmienia coś poza zwracanym wynikiem.

Przykłady:

- zapis do pliku,
- modyfikacja globalnej zmiennej,
- wysłanie danych przez sieć,
- modyfikacja przekazanego obiektu,
- wypisanie na ekran.

Efekty uboczne są potrzebne, ale warto je kontrolować.

## Funkcja czysta

W uproszczeniu funkcja czysta:

- dla tych samych argumentów daje ten sam wynik,
- nie zmienia zewnętrznego stanu.

Przykład:

````python
def calculate_average(values):
    return sum(values) / len(values)

print(calculate_average([1, 2, 3]))
````

## Docstring jako kontrakt

Docstring powinien pomóc odpowiedzieć:

- co funkcja robi,
- czego oczekuje,
- co zwraca,
- jakie wyjątki może zgłosić.

Nie powinien tylko przepisywać nazwy funkcji.

````python
def celsius_to_kelvin(value: float) -> float:
    """Convert Celsius temperature to Kelvin.

    Raises:
        ValueError: If value is below absolute zero.
    """
    if value < -273.15:
        raise ValueError("Temperature below absolute zero")

    return value + 273.15
````

## Type hints

Adnotacje typów:

```python

```

pomagają:

- czytelnikowi,
- IDE,
- narzędziom analizy statycznej.

Nie są domyślnie pełnym mechanizmem kontroli typów w czasie działania.


## `assert`

`assert` pozwala sprawdzić założenie programisty:

```python
assert condition
```

Jeśli warunek jest fałszywy:

```text
AssertionError
```

````python
def average(values):
    assert len(values) > 0
    return sum(values) / len(values)

print(average([1, 2, 3]))
````

## `assert` nie służy do walidacji danych użytkownika

Nie używamy:

```python
assert user_age >= 0
```

jako mechanizmu kontroli danych wejściowych.

Dla danych użytkownika używamy normalnych warunków i wyjątków.

`assert` służy głównie do sprawdzania założeń wewnętrznych programu.

## Test jednostkowy

Test jednostkowy sprawdza mały fragment programu — najczęściej funkcję lub metodę.

Schemat:

```text
ARRANGE
↓
przygotuj dane

ACT
↓
wykonaj operację

ASSERT
↓
sprawdź wynik
```

## Najprostszy test przez `assert`

Dla funkcji:

```python
def add(a, b):
    return a + b
```

test:

```python
assert add(2, 3) == 5
```

To już jest automatyczna weryfikacja zachowania.

````python
def add(a, b):
    return a + b

assert add(2, 3) == 5
assert add(-1, 1) == 0

print("Testy zakończone")
````

## Test powinien sprawdzać przypadek, nie implementację

Nie interesuje nas, **jak** funkcja liczy.

Interesuje nas kontrakt:

```text
wejście
↓
oczekiwany wynik
```

Dzięki temu implementację można zmieniać bez przepisywania testów.

## Przypadki typowe i brzegowe

Dla:

```python
celsius_to_kelvin(value)
```

warto sprawdzić:

```text
0 °C
100 °C
-273.15 °C
wartość niższą od -273.15 °C
```

Błędy często ukrywają się na granicach.

## Testowanie wyjątku

Możemy jawnie sprawdzić, że funkcja zgłasza oczekiwany wyjątek:

````python
try:
    celsius_to_kelvin(-300)
except ValueError:
    print("Test zaliczony")
else:
    raise AssertionError(
        "Oczekiwano ValueError"
    )
````

## `unittest` — biblioteka standardowa

Python zawiera framework testowy:

```python
import unittest
```

Przykład:

````python
import unittest


class TestTemperature(unittest.TestCase):

    def test_zero_celsius(self):
        self.assertEqual(
            celsius_to_kelvin(0),
            273.15,
        )

    def test_invalid_temperature(self):
        with self.assertRaises(ValueError):
            celsius_to_kelvin(-300)
````

## Testy regresyjne

Gdy znajdziemy błąd:

1. tworzymy test, który go odtwarza,
2. test powinien najpierw nie przejść,
3. poprawiamy kod,
4. test zaczyna przechodzić.

Od tej pory test pilnuje, aby ten błąd nie wrócił.

## Test w stylu pytest

```python
def test_zero_celsius():
    assert celsius_to_kelvin(0) == 273.15
```

Framework automatycznie wykrywa funkcje testowe i prezentuje czytelny raport.

## Testy parametryzowane — idea

Ta sama reguła może być sprawdzana dla wielu danych:

```text
wejście 0       → 273.15
wejście 100     → 373.15
wejście -40     → 233.15
```

Nie warto ręcznie kopiować identycznej struktury testu.

## Dane testowe w pętli

Bez dodatkowych bibliotek:

````python
cases = [
    (0, 273.15),
    (100, 373.15),
    (-40, 233.15),
]

for value, expected in cases:
    result = celsius_to_kelvin(value)

    assert result == expected
````

## Formatter

Formatter automatycznie ustala styl kodu.

Jego zadaniem nie jest znalezienie błędu logicznego, tylko ujednolicenie zapisu.

Dzięki temu zespół nie traci czasu na dyskusje o odstępach i łamaniu linii.

## Linter

Linter analizuje kod i wskazuje m.in.:

- nieużywane importy,
- podejrzane konstrukcje,
- naruszenia konwencji,
- część potencjalnych błędów.

Przykładem współczesnego narzędzia jest `ruff`.

## Type checker

Narzędzie takie jak `mypy` może analizować adnotacje typów bez uruchamiania programu.

Przykład problemu:

```python
def add(a: int, b: int) -> int:
    return a + b

add("2", 3)
```

Python może dojść do tej linii dopiero podczas wykonania, a analiza statyczna może ostrzec wcześniej.

## Trzy różne narzędzia

## Przykład: kod łatwy do testowania

```python
def is_alarm(value: float, limit: float = 30.0) -> bool:
    return value > limit
```

Nie czyta pliku, nie używa globalnego stanu i nie wypisuje.

Dzięki temu test jest banalny.

````python
def is_alarm(
    value: float,
    limit: float = 30.0,
) -> bool:
    return value > limit


assert is_alarm(31)
assert not is_alarm(30)
assert not is_alarm(29)
````
