## Klasa

Klasa opisuje nowy typ obiektów.

Najprostsza definicja:

```python
class Sensor:
    pass
```

Na razie klasa niczego szczególnego nie robi, ale możemy już tworzyć jej obiekty.

````python
class Sensor:
    pass

sensor_1 = Sensor()
sensor_2 = Sensor()

print(sensor_1)
print(sensor_2)
````

Jedna klasa może być podstawą do utworzenia wielu niezależnych obiektów.

## Atrybuty obiektu

Obiekt może przechowywać własne dane:

```python
sensor_1.name = "sensor_01"
sensor_1.unit = "C"
```

Te dane nazywamy **atrybutami**.

````python
sensor_1.name = "sensor_01"
sensor_1.unit = "C"

sensor_2.name = "sensor_02"
sensor_2.unit = "%"

print(sensor_1.name, sensor_1.unit)
print(sensor_2.name, sensor_2.unit)
````

## Problem z ręcznym tworzeniem atrybutów

Ręczne przypisywanie:

```python
sensor.name = ...
sensor.unit = ...
```

jest podatne na błędy.

Możemy zapomnieć utworzyć atrybut albo nadać mu inną nazwę.

Dlatego definiujemy sposób inicjalizacji obiektu.

## `__init__`

```python
class Sensor:

    def __init__(self, name, unit):
        self.name = name
        self.unit = unit
```

Metoda `__init__` służy do **inicjalizacji już utworzonej instancji**.

## Ważna terminologia

Często mówi się, że `__init__` jest konstruktorem.

Dokładniej:

> `__init__` jest metodą inicjalizującą obiekt.

Samo tworzenie instancji zaczyna się wcześniej.

Na tym etapie nie musimy jeszcze analizować mechanizmu `__new__`.

````python
class Sensor:

    def __init__(self, name, unit):
        self.name = name
        self.unit = unit

sensor_1 = Sensor("sensor_01", "C")
sensor_2 = Sensor("sensor_02", "%")

print(sensor_1.name, sensor_1.unit)
print(sensor_2.name, sensor_2.unit)
````

## Co oznacza `self`?

`self` oznacza:

> obiekt, na którym aktualnie wykonywana jest metoda.

Dla:

```python
sensor_1 = Sensor("sensor_01", "C")
```

w trakcie `__init__`:

```text
self
↓
sensor_1
```

## `self.name = name`

Porównaj:

```python
def __init__(self, name):
    self.name = name
```

Po prawej:

```text
name
→ parametr funkcji
```

Po lewej:

```text
self.name
→ atrybut obiektu
```

To dwa różne poziomy nazw.

## Metoda

Metoda to funkcja zdefiniowana wewnątrz klasy.

Przykład:

````python
class Sensor:

    def __init__(self, name, unit):
        self.name = name
        self.unit = unit

    def show(self):
        print(f"{self.name}: jednostka = {self.unit}")

sensor = Sensor("sensor_01", "C")
sensor.show()
````

## Co dzieje się przy `sensor.show()`?

Zapis:

```python
sensor.show()
```

możemy intuicyjnie rozumieć podobnie do:

```python
Sensor.show(sensor)
```

Obiekt jest przekazywany do parametru `self`.

````python
Sensor.show(sensor)
````

## Stan obiektu

Dodajmy bieżący pomiar:

```python
self.value = None
```

Stan obiektu może się zmieniać podczas działania programu.

````python
class Sensor:

    def __init__(self, name, unit):
        self.name = name
        self.unit = unit
        self.value = None

sensor = Sensor("sensor_01", "C")

print(sensor.value)
````

## Metoda może zmieniać stan

```python
def update(self, value):
    self.value = value
```

Po wywołaniu metoda modyfikuje atrybut obiektu.

````python
class Sensor:

    def __init__(self, name, unit):
        self.name = name
        self.unit = unit
        self.value = None

    def update(self, value):
        self.value = value

sensor = Sensor("sensor_01", "C")

print(sensor.value)

sensor.update(21.5)
print(sensor.value)

sensor.update(35.2)
print(sensor.value)
````

## Stan i zachowanie

To najważniejszy model tego wykładu:

```text

↓

ZACHOWANIE
↓

```

Metoda może:

- odczytać stan,
- wykonać obliczenie,
- zmienić stan.

## Zachowanie zależne od stanu

Dodajmy metodę alarmową:

````python
class Sensor:

    def __init__(self, name, unit):
        self.name = name
        self.unit = unit
        self.value = None

    def update(self, value):
        self.value = value

    def is_alarm(self, limit=30):
        return (
            self.value is not None
            and self.value > limit
        )

sensor = Sensor("sensor_01", "C")

print(sensor.is_alarm())

sensor.update(35.2)

print(sensor.is_alarm())
````

## Obiekt może pilnować własnych reguł

Dla czujnika temperatury fizycznie niemożliwa wartość:

```text
-300 °C
```

nie powinna zostać zaakceptowana.

Możemy umieścić tę regułę w metodzie `update()`.

````python
class TemperatureSensor:

    def __init__(self, name, unit="C"):
        self.name = name
        self.unit = unit
        self.value = None

    def update(self, value):
        if value < -273.15:
            raise ValueError("Nieprawidłowy pomiar temperatury")

        self.value = value

sensor = TemperatureSensor("sensor_01")

sensor.update(21.7)
print(sensor.value)
````

## Odpowiedzialność obiektu

Obiekt może nie tylko przechowywać dane.

Może również pilnować, aby jego stan był poprawny.

```text

↓
jedna klasa
```

To jedna z głównych korzyści dobrze zaprojektowanego OOP.

## Atrybut instancji

Dlatego dwa obiekty mogą mieć różne wartości:

## Atrybut klasy

Możemy również utworzyć nazwę bezpośrednio w klasie:

```python
class Sensor:
    category = "measurement"
```

Taki atrybut należy do klasy i może być współdzielony przez jej instancje.

````python
class Sensor:

    category = "measurement"

    def __init__(self, name):
        self.name = name

sensor_1 = Sensor("A")
sensor_2 = Sensor("B")

print(sensor_1.category)
print(sensor_2.category)
print(Sensor.category)
````

## Tożsamość obiektu

Dwa obiekty tej samej klasy nie muszą być tym samym obiektem.

## `is` a `==`

`is` pyta:

> czy to dokładnie ten sam obiekt?

`==` pyta:

> czy obiekty są równe według reguł danego typu?

Dla własnej klasy, bez dodatkowej definicji równości, wynik może być inny niż intuicyjnie oczekujemy.

````python
sensor_1 = Sensor("A")
sensor_2 = Sensor("A")

print(sensor_1 is sensor_2)
print(sensor_1 == sensor_2)
````

## `id()`

`id()` zwraca identyfikator obiektu w czasie jego życia:

````python
sensor_1 = Sensor("A")
sensor_2 = Sensor("A")
sensor_3 = sensor_1

print(id(sensor_1))
print(id(sensor_2))
print(id(sensor_3))
````

## Publiczny atrybut

W Pythonie:

## Konwencja z pojedynczym `_`

```python
self._value
```

oznacza umownie:

> ten element jest częścią implementacji wewnętrznej.

Python nadal pozwala go odczytać.

To **konwencja**, a nie mechanizm kontroli dostępu.

## Podwójne `__`

```python
self.__value
```

uruchamia mechanizm **name mangling**.

Python zmienia wewnętrzną nazwę atrybutu, aby ograniczyć przypadkowe kolizje nazw.

To nie jest pełna prywatność znana np. z Javy czy C++.

````python
class Example:

    def __init__(self):
        self.public = 1
        self._internal = 2
        self.__mangled = 3

example = Example()

print(example.public)
print(example._internal)

# Nie zadziała:
# print(example.__mangled)

print(example.__dict__)
````

## Python nie wymaga getterów i setterów dla wszystkiego

Nie musimy pisać:

```python
get_value()
set_value(...)
```

dla każdego atrybutu.

Naturalny Pythonowy interfejs może wyglądać po prostu:

```python
sensor.value
```

Dopiero gdy pojawia się dodatkowa logika, możemy użyć `property`.

## `property`

`property` pozwala zachować składnię atrybutu:

```python
sensor.value
```

a jednocześnie wykonywać kod przy jego odczycie lub zmianie.

````python
class TemperatureSensor:

    def __init__(self, name):
        self.name = name
        self._value = None

    @property
    def value(self):
        return self._value

sensor = TemperatureSensor("sensor_01")

print(sensor.value)
````

## Setter właściwości

Możemy dodać walidację przy przypisaniu:

````python
class TemperatureSensor:

    def __init__(self, name):
        self.name = name
        self._value = None

    @property
    def value(self):
        return self._value

    @value.setter
    def value(self, new_value):
        if new_value < -273.15:
            raise ValueError("Nieprawidłowa temperatura")

        self._value = new_value

sensor = TemperatureSensor("sensor_01")

sensor.value = 21.7

print(sensor.value)
````

## Co daje `property`?

Użytkownik klasy nadal pisze:

```python
sensor.value = 21.7
```

ale klasa może:

- sprawdzić poprawność wartości,
- przeliczyć dane,
- kontrolować sposób odczytu lub zapisu.

Interfejs pozostaje prosty.

## Nie musimy używać `property` wszędzie

Jeżeli:

```python
self.name = name
```

nie wymaga żadnej dodatkowej logiki, zwykły publiczny atrybut jest często najlepszym rozwiązaniem.

Nie komplikujemy klasy bez potrzeby.

## Przykład integrujący

Zbudujmy klasę reprezentującą czujnik temperatury:

````python
class TemperatureSensor:

    category = "temperature"

    def __init__(self, name, unit="C"):
        self.name = name
        self.unit = unit
        self._value = None

    @property
    def value(self):
        return self._value

    def update(self, value):
        if value < -273.15:
            raise ValueError("Nieprawidłowy pomiar")

        self._value = value

    def is_alarm(self, limit=30):
        return (
            self._value is not None
            and self._value > limit
        )
````

````python
sensor = TemperatureSensor("sensor_01")

sensor.update(35.2)

print("name:", sensor.name)
print("category:", sensor.category)
print("value:", sensor.value)
print("alarm:", sensor.is_alarm())
````

## OOP nie zastępuje wszystkich innych technik

Klasa jest dobrym rozwiązaniem, gdy:

- dane tworzą logiczną całość,

Nie każda funkcja musi stać się metodą i nie każdy słownik musi stać się klasą.
