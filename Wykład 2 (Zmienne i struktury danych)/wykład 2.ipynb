## Nazwa i obiekt

W Pythonie wygodniej myśleć:

```text
nazwa                    obiekt
temperature ───────────► 21.5
                           float
```

Instrukcja:

```python
temperature = 21.5
```

wiąże nazwę `temperature` z obiektem reprezentującym wartość `21.5`.

> Typ jest właściwością obiektu, a nie nazwy.

````python
temperature = 21.5

print(temperature)
print(type(temperature))
````

## Dynamiczny system typów

Ta sama nazwa może w różnych momentach wskazywać na obiekty różnych typów.

````python
value = 10
print(value, type(value))

value = 10.5
print(value, type(value))

value = "AGH"
print(value, type(value))
````

## Podstawowe typy pojedynczych wartości

| Przykład | Typ | Znaczenie |
|---|---|---|
| `42` | `int` | liczba całkowita |
| `3.14` | `float` | liczba zmiennoprzecinkowa |
| `True` | `bool` | wartość logiczna |
| `"Python"` | `str` | napis |

Python posiada również inne typy, np. `complex`, ale nie będą one dzisiaj głównym tematem.

````python
integer_value = 42
float_value = 3.14
bool_value = True
text_value = "Python"
missing_value = None

print(type(integer_value))
print(type(float_value))
print(type(bool_value))
print(type(text_value))
print(type(missing_value))
````

## Kolekcje

Pojedyncza wartość często nie wystarcza.

Python udostępnia kilka podstawowych sposobów przechowywania wielu obiektów:

```text
list   — uporządkowana i mutowalna
tuple  — uporządkowana i niemutowalna
dict   — pary klucz → wartość
set    — unikalne elementy
```

Każda z tych struktur odpowiada na inne potrzeby.

## Lista — `list`

Lista jest uporządkowaną, **mutowalną** kolekcją.

Typowy przykład:

```python
temperatures = [21.5, 22.1, 19.8, 20.4]
```

Python pozwala przechowywać w liście obiekty różnych typów, choć w praktyce często grupujemy dane tego samego rodzaju.

## Tworzenie list

Najczęstszy sposób — literał listy:

```python
numbers = [1, 2, 3, 4]
names = ["Anna", "Jan", "Maria"]
```

Pusta lista:

```python
values = []
values = list()
```

Lista może również powstać z innego obiektu iterowalnego:

```python
letters = list("Python")
numbers = list(range(5))
```

````python
numbers = [1, 2, 3, 4]
empty_1 = []
empty_2 = list()
letters = list("Python")
from_range = list(range(5))

print(numbers)
print(empty_1, empty_2)
print(letters)
print(from_range)
````

## Dostęp do elementów listy

Indeksy zaczynają się od zera.

```text
[10, 20, 30, 40]
  0   1   2   3
 -4  -3  -2  -1
```

Wycinek ma postać:

```python
lista[start:stop]
```

Element o indeksie `start` należy do wyniku, a element `stop` już nie.

````python
temperatures = [21.5, 22.1, 19.8, 20.4]

print(temperatures[0])
print(temperatures[-1])
print(temperatures[1:3])
print(temperatures[:2])
print(temperatures[2:])
````

## Lista jest mutowalna

Po utworzeniu listy możemy zmieniać jej elementy i dodawać nowe.

````python
temperatures = [21.5, 22.1, 19.8]

temperatures[1] = 23.0
temperatures.append(20.4)

print(temperatures)
````

## Ten sam rezultat, różny zapis

Pustą listę możemy utworzyć na co najmniej dwa sposoby:

```python
[]
list()
```

Czy Python wykonuje dokładnie tę samą pracę?

Możemy to sprawdzić eksperymentalnie.

````python
%timeit []
%timeit list()
````

## Co pokazuje pomiar czasu?

Zwykle literał:

```python
[]
```

jest szybszy niż:

```python
list()
```

ponieważ drugi zapis wymaga dodatkowego wywołania konstruktora.

### Ale uwaga

Nie chodzi o optymalizowanie pojedynczej pustej listy.

Ważniejsza obserwacja brzmi:

> Dwa zapisy dające podobny wynik mogą uruchamiać różne mechanizmy interpretera.

## Lista zagnieżdżona — ważna pułapka

Na pierwszy rzut oka ten zapis wygląda dobrze:

```python
matrix = [[0] * 3] * 3
```

Otrzymamy strukturę przypominającą macierz 3 × 3.

Czy jednak powstały trzy niezależne listy?

````python
matrix = [[0] * 3] * 3

print(matrix)

matrix[0][0] = 1

print(matrix)
````

## Dlaczego zmieniły się trzy wiersze?

Operator `*` powielił **referencję do tej samej listy wewnętrznej**.

Poprawna konstrukcja niezależnych wierszy może wyglądać tak:

```python
matrix = [[0] * 3 for _ in range(3)]
```

Wrócimy później do składni `for` i list comprehensions.

### Wniosek

Kod może wyglądać podobnie i nawet dawać podobny początkowy wynik, a mimo to mieć inną strukturę obiektów w pamięci.

````python
matrix = [[0] * 3 for _ in range(3)]

matrix[0][0] = 1

print(matrix)
````

## Krotka — `tuple`

Krotka jest uporządkowaną, **niemutowalną** kolekcją.

Typowe zastosowania:

- współrzędne,
- stałe zestawy wartości,
- zwracanie kilku wartości razem.

## Tworzenie krotek

Najczęściej:

```python
point = (10, 20)
person = ("Anna", "Kowalska", 20)
```

Pusta krotka:

```python
empty = ()
empty = tuple()
```

Jednoelementowa krotka wymaga przecinka:

```python
single = (10,)
```

````python
a = (10)
b = (10,)

print(a, type(a))
print(b, type(b))
````

## To przecinek tworzy krotkę

Nawiasy często poprawiają czytelność, ale nie są najważniejszym elementem składni.

Ten zapis również tworzy krotkę:

```python
point = 10, 20
```

Przyda się to później przy pakowaniu i rozpakowywaniu wartości.

````python
point = 10, 20

print(point)
print(type(point))
````

## Dostęp do elementów krotki

Indeksowanie i slicing działają podobnie jak dla list.

````python
point = (10, 20, 30)

print(point[0])
print(point[-1])
print(point[1:])
````

## Krotka jest niemutowalna

Po utworzeniu nie możemy podmienić jej elementu.

````python
point = (10, 20)

# Odkomentuj, aby zobaczyć TypeError:
# point[0] = 15

print(point)
````

## Czas inicjalizacji pustej krotki

Podobnie jak dla listy możemy porównać:

```python
()
tuple()
```

````python
%timeit ()
%timeit tuple()
````

## Słownik — `dict`

Słownik przechowuje pary:

```text
klucz → wartość
```

Przykład:

```python
student = {
    "name": "Anna",
    "surname": "Kowalska",
    "age": 20
}
```

Każdy klucz w słowniku musi być unikalny.

## Tworzenie słowników

Literał:

```python
student = {
    "name": "Anna",
    "age": 20
}
```

Pusty słownik:

```python
student = {}
student = dict()
```

Możliwa jest też konstrukcja:

```python
student = dict(name="Anna", age=20)
```

````python
student_1 = {
    "name": "Anna",
    "age": 20
}

student_2 = dict(
    name="Anna",
    age=20
)

empty_1 = {}
empty_2 = dict()

print(student_1)
print(student_2)
print(type(empty_1), type(empty_2))
````

## Słownik można utworzyć z par

`dict()` potrafi zbudować słownik z kolekcji par:

```python
pairs = [
    ("name", "Anna"),
    ("age", 20)
]

student = dict(pairs)
```

````python
pairs = [
    ("name", "Anna"),
    ("age", 20)
]

student = dict(pairs)

print(student)
````

## Dostęp i modyfikacja słownika

Do wartości odwołujemy się za pomocą **klucza**, nie indeksu pozycyjnego.

````python
student = {
    "name": "Anna",
    "surname": "Kowalska",
    "age": 20
}

print(student["name"])

student["age"] = 21
student["group"] = "1A"

print(student)
print(student.keys())
print(student.values())
````

## Jakie obiekty mogą być kluczami?

Klucz słownika musi być **hashowalny**.

Na tym etapie wystarczy zapamiętać:

- `str` — tak,
- `int` — tak,
- odpowiednia `tuple` — tak,
- `list` — nie.

Później dokładniej omówimy, dlaczego mutowalne obiekty zwykle nie nadają się na klucze.

````python
data = {
    "name": "Anna",
    1: "jeden",
    (10, 20): "punkt"
}

print(data)
````

## Uwaga: `(4)` i `(4,)`

```python

```

to po prostu liczba `4`.

Jednoelementowa krotka to:

```python
(4,)
```

Ma to znaczenie również wtedy, gdy używamy krotki jako klucza słownika.

````python
print(type((4)))
print(type((4,)))
````

## Zbiór — `set`

Zbiór przechowuje **unikalne** elementy.

Nie odwołujemy się do elementów przez indeks pozycyjny.

`set` jest **mutowalny** — po utworzeniu możemy dodawać i usuwać elementy.

````python
numbers = {1, 1, 2, 2, 3, 3}

print(numbers)

numbers.add(4)
numbers.remove(2)

print(numbers)
````

## Tworzenie zbiorów

Niepusty zbiór:

```python
numbers = {1, 2, 3}
```

Zbiór z istniejącej kolekcji:

```python
numbers = set([1, 2, 3, 2])
```

### Pułapka

```python
{}
```

tworzy pusty **słownik**, a nie pusty zbiór.

Pusty zbiór tworzymy:

```python
set()
```

````python
a = {}
b = set()

print(type(a))
print(type(b))
````

## Zbiór jako sposób usuwania duplikatów

Konstruktor `set(...)` jest przydatny, gdy dane już istnieją w innej kolekcji.

````python
measurements = [1, 2, 2, 3, 3, 3]

unique_measurements = set(measurements)

print(measurements)
print(unique_measurements)
````

## Nie każdy obiekt może być elementem zbioru

Element `set` musi być hashowalny.

Lista jest mutowalna, dlatego nie może być elementem zbioru.

````python
valid_set = {1, 2, (3, 4)}
print(valid_set)

# Odkomentuj, aby zobaczyć TypeError:
# invalid_set = {1, 2, [3, 4]}
````

## Różne sposoby inicjalizacji zbioru

Porównanie:

```python
{1, 2, 3}
set([1, 2, 3])
```

Oba zapisy mogą dać ten sam zbiór, ale wykonują inne operacje.

Pierwszy jest literałem, drugi konstruuje zbiór na podstawie istniejącego obiektu iterowalnego.

````python
%timeit {1, 2, 3}
%timeit set([1, 2, 3])
````

## Jak interpretować wyniki `timeit`?

Wyniki zależą od:

- wersji Pythona,
- implementacji interpretera,
- systemu operacyjnego,
- sprzętu,
- aktualnego obciążenia komputera.

Dlatego nie interesuje nas pojedyncza liczba nanosekund.

Interesuje nas przede wszystkim:

> **dlaczego dwa zapisy mogą mieć różny koszt wykonania?**

Kolejność priorytetów na tym kursie:

**poprawność → czytelność → właściwa struktura → wydajność**.
