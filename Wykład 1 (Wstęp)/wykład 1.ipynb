## Programowanie zaczyna się od problemu

Programista nie zaczyna od pisania kodu.

Najpierw istnieje **problem**, który chcemy rozwiązać.

Przykład:

> Mamy trzy pomiary temperatury i chcemy obliczyć ich średnią.

To jeszcze nie jest program.

## Od problemu do algorytmu

**Algorytm** to uporządkowany sposób rozwiązania problemu.

Dla naszego przykładu:

1. Pobierz trzy wartości temperatury.
2. Dodaj je.
3. Podziel sumę przez liczbę pomiarów.
4. Wyświetl wynik.

Algorytm można opisać:
- słowami,
- pseudokodem,
- schematem blokowym,
- kodem programu.

````python
t1 = 21.5
t2 = 22.3
t3 = 20.8

average = (t1 + t2 + t3) / 3

print(average)
````

## Problem → algorytm → program → wynik

```text
PROBLEM
   ↓
ALGORYTM
   ↓
PROGRAM
   ↓
KOMPUTER
   ↓
WYNIK
```

Kod źródłowy jest **formalnym zapisem rozwiązania**, który komputer może wykonać.

Później zobaczymy konstrukcje, które tę kolejność zmieniają:
- instrukcje warunkowe,
- pętle,
- funkcje.

````python
print("A")
print("B")
print("C")
````

## Język programowania

Język programowania jest formalnym językiem służącym do zapisywania instrukcji wykonywanych przez komputer.

Każdy język określa między innymi:

- dozwolone słowa i symbole,
- reguły składni,
- znaczenie poszczególnych konstrukcji.

Kod musi być zapisany zgodnie z tymi regułami, aby narzędzie wykonujące program mogło go zrozumieć.

## Języki niskiego i wysokiego poziomu

### Języki niskiego poziomu

Są blisko sposobu działania procesora i sprzętu.

### Języki wysokiego poziomu

Pozwalają zapisywać rozwiązanie w sposób znacznie łatwiejszy do zrozumienia przez człowieka.

Python jest językiem **wysokiego poziomu**.

## Czym jest Python?

Python jest:

- językiem programowania wysokiego poziomu,
- językiem ogólnego przeznaczenia,
- językiem o dynamicznym systemie typów,
- językiem wieloparadygmatowym,
- projektem otwartym (open source).

Jedną z jego najważniejszych cech jest nacisk na **czytelność kodu**.

Dokumentacja: [docs.python.org](https://docs.python.org/3/)

## Dlaczego Python?

Python jest dobrym językiem do nauki programowania, ponieważ:

- ma stosunkowo czytelną składnię,

- posiada rozbudowaną bibliotekę standardową,
- ma bardzo duży ekosystem dodatkowych bibliotek,

## Python nie jest najlepszy do wszystkiego

Wybór języka programowania jest decyzją inżynierską.

Python nie zawsze będzie pierwszym wyborem np. dla:

- bardzo ograniczonych mikrokontrolerów,
- fragmentów systemów o bardzo ostrych wymaganiach czasu rzeczywistego,
- aplikacji wymagających maksymalnej wydajności pojedynczych operacji,

W praktyce system może wykorzystywać **więcej niż jeden język programowania**.

## Wejście → przetwarzanie → wyjście

Bardzo wiele programów można na początku rozumieć za pomocą prostego modelu:

```text
WEJŚCIE
   ↓
PRZETWARZANIE
   ↓
WYJŚCIE
```

Przykład:

- wejście — temperatura w °C,
- przetwarzanie — przeliczenie jednostki,
- wyjście — temperatura w °F.

````python
temperature_c = float(input("Podaj temperaturę [°C]: "))

temperature_f = temperature_c * 9 / 5 + 32

print(f"{temperature_c} °C = {temperature_f} °F")
````

## Błędy są częścią programowania

Podczas pisania programu będziemy popełniać błędy.

To normalny element pracy programisty.

Najważniejsze jest nauczenie się:

1. zauważenia błędu,
2. przeczytania komunikatu,
3. znalezienia miejsca, którego dotyczy,
4. ustalenia przyczyny,
5. poprawienia i ponownego uruchomienia programu.

## Przykład: błąd składni

Ten zapis nie jest zgodny z gramatyką języka:

```python
print("Witaj"
```

Python zgłosi:

```text
SyntaxError
```

**SyntaxError** oznacza, że interpreter nie potrafi poprawnie odczytać struktury kodu.

## Przykład: nieznana nazwa

```python
print(not_defined_variable)
```

Python zgłosi:

```text
NameError
```

Program jest składniowo poprawny, ale używa nazwy, która nie została wcześniej zdefiniowana.

## Jak czytać komunikat błędu?

Na początku szukaj trzech informacji:

```text
GDZIE?
↓
która linia programu

CO?
↓
typ błędu

DLACZEGO?
↓
opis błędu
```

Nie trzeba od razu rozumieć całego tracebacka.

Z czasem stanie się jednym z podstawowych narzędzi pracy.
