```python
True

```

Warunki w programie ostatecznie prowadzą do odpowiedzi:

## Truthiness — prawdziwość obiektów

W Pythonie wiele obiektów może być użytych bezpośrednio jako warunek.

Zwykle jako `False` traktowane są m.in.:

```python
0
0.0
""
[]
{}
set()
None
```

Większość innych wartości jest traktowana jako `True`.

````python
values = [
    0,
    1,
    "",
    "Python",
    [],
    [1, 2],
    {},
    {"a": 1},
    None,
]

for value in values:
    print(repr(value), "->", bool(value))
````

## Po co nam truthiness?

Zamiast:

```python
measurements = []

if len(measurements) > 0:
    print("Są pomiary")
```

często piszemy po prostu:

```python
if measurements:
    print("Są pomiary")
```

Kod jest krótszy i naturalny dla Pythona.

## Operatory porównania

| Operator | Znaczenie |
|:---:|---|
| `==` | równe |
| `!=` | różne |
| `<` | mniejsze |
| `<=` | mniejsze lub równe |
| `>` | większe |
| `>=` | większe lub równe |

````python
temperature = 28

print(temperature == 28)
print(temperature != 20)
print(temperature > 30)
print(temperature <= 30)
````

## Łączenie warunków

Do budowania bardziej złożonych warunków używamy:

- `and` — oba warunki muszą być prawdziwe,
- `or` — wystarczy jeden prawdziwy,
- `not` — zaprzeczenie.

````python
temperature = 25
active = True

print(temperature > 20 and active)
print(temperature > 30 or active)
print(not active)
````

## Porównania łańcuchowe

Python pozwala pisać warunki podobnie jak w matematyce:

```python
18 <= age < 26
```

zamiast:

```python
age >= 18 and age < 26
```

Oba zapisy są poprawne, ale pierwszy jest zwykle czytelniejszy.

````python
age = 20

print(18 <= age < 26)
print(age >= 18 and age < 26)
````

## Operator `in`

`in` sprawdza, czy element należy do kolekcji.

````python
sensors = ["temperature", "pressure", "humidity"]

print("temperature" in sensors)
print("speed" in sensors)
print("speed" not in sensors)
````

## `==` a `is`

To dwa różne pytania:

```text
==  → czy wartości są równe?
is  → czy to dokładnie ten sam obiekt?
```

````python
a = [1, 2]
b = [1, 2]
c = a

print(a == b)
print(a is b)

print(a == c)
print(a is c)
````

## Typowe użycie `is`: porównanie z `None`

Dla `None` używamy:

```python
result is None
```

lub:

```python
result is not None
```

Nie uczymy się `is` jako zamiennika dla `==`.

````python
result = None

if result is None:
    print("Brak wyniku")
````

## Instrukcje sterujące

Najważniejsze konstrukcje sterujące w tym wykładzie:

- `if` — wybór ścieżki wykonania,
- `match` — dopasowanie wzorca lub wartości,
- `while` — powtarzanie zależne od warunku,

## Instrukcja `if`

Najprostsza postać:

```python
if warunek:
    instrukcje
```

Blok kodu jest wykonywany tylko wtedy, gdy warunek jest prawdziwy.

````python
temperature = 35

if temperature > 30:
    print("Wysoka temperatura")

print("Koniec programu")
````

## `if ... else`

Jeżeli chcemy obsłużyć dwa przypadki:

```python
if warunek:
    ...
else:
    ...
```

````python
temperature = 18

if temperature >= 20:
    print("Ciepło")
else:
    print("Chłodno")
````

## `if ... elif ... else`

Gdy możliwych wariantów jest więcej:

```python
if warunek_1:
    ...
elif warunek_2:
    ...
else:
    ...
```

Warunki są sprawdzane **od góry do dołu**.

````python
temperature = 27

if temperature >= 30:
    print("Gorąco")
elif temperature >= 20:
    print("Ciepło")
elif temperature >= 10:
    print("Chłodno")
else:
    print("Zimno")
````

## Kolejność warunków ma znaczenie

Rozważ:

```python
temperature = 35

if temperature >= 20:
    print("Ciepło")
elif temperature >= 30:
    print("Gorąco")
```

Drugi warunek nie zostanie już sprawdzony.

### Zasada

W gałęziach `if / elif / else` wykonuje się pierwsza pasująca gałąź.

## Zagnieżdżanie `if`

Warunki mogą być zagnieżdżane, ale zbyt wiele poziomów pogarsza czytelność.

````python
number = -8

if number >= 0:
    if number == 0:
        print("Zero")
    else:
        print("Liczba dodatnia")
else:
    print("Liczba ujemna")
````

## `match` — prosty przykład

`match` jest dostępny od Pythona 3.10.

Dla prostych poleceń może zastąpić długi ciąg `if / elif`.

````python
command = "start"

match command:
    case "start":
        print("Uruchamiam")
    case "stop":
        print("Zatrzymuję")
    case "status":
        print("Pokazuję stan")
    case _:
        print("Nieznane polecenie")
````

## `match` potrafi więcej niż klasyczny `switch`

Python wspiera **structural pattern matching**.

Przykład:

```python
point = (0, 10)

match point:
    case (0, 0):
        print("Początek układu")
    case (0, y):
        print("Punkt na osi Y")
    case (x, y):
        print("Zwykły punkt")
```

Na tym etapie traktujemy to jako rozszerzenie — nie trzeba jeszcze znać wszystkich wzorców.

## Pętla `while`

`while` oznacza:

> wykonuj blok dopóki warunek jest prawdziwy.

````python
counter = 0

while counter < 5:
    print(counter)
    counter += 1
````

## Uwaga na pętlę nieskończoną

Jeżeli warunek nigdy nie stanie się fałszywy:

```python
counter = 0

while counter < 5:
    print(counter)
```

pętla nie zakończy się sama.

W `while` musimy rozumieć, **co zmienia stan warunku**.

## Ten sam problem z `while`

Chcemy wypisać wszystkie temperatury z listy:

````python
temperatures = [21.5, 22.1, 20.4]

i = 0

while i < len(temperatures):
    print(temperatures[i])
    i += 1
````

## Pętla `for`

W Pythonie `for` przede wszystkim **iteruje po elementach obiektu**.

To bardzo ważne:

> nie myśl „for = licznik”, tylko „for = kolejne elementy”.

````python
temperatures = [21.5, 22.1, 20.4]

for temperature in temperatures:
    print(temperature)
````

## `while` czy `for`?

```text
while
→ powtarzaj, dopóki warunek jest prawdziwy

for
→ wykonaj coś dla kolejnych elementów
```

W wielu sytuacjach oba podejścia są możliwe, ale jedno z nich będzie lepiej opisywać intencję programu.

## Funkcja `range()`

`range()` nie tworzy „wektora liczb”.

Tworzy obiekt reprezentujący zakres liczb całkowitych.

Najczęstsze formy:

```python
range(stop)
range(start, stop)
range(start, stop, step)
```

````python
r = range(5)

print(r)
print(type(r))
print(list(r))
````

````python
print(list(range(5)))
print(list(range(2, 8)))
print(list(range(2, 10, 2)))
print(list(range(10, 2, -2)))
````

## Kiedy używać `range()`?

Gdy rzeczywiście potrzebujemy licznika lub powtórzenia określoną liczbę razy.

````python
for i in range(5):
    print("Iteracja:", i)
````

## Iterowanie po różnych obiektach

W Pythonie wiele obiektów jest **iterowalnych**.

````python
for char in "AGH":
    print(char)

for value in [10, 20, 30]:
    print(value)

for key in {"a": 1, "b": 2}:
    print(key)
````

## Dwie wersje tego samego zadania

Wersja indeksowa:

```python
for i in range(len(values)):
    print(values[i])
```

Wersja bezpośrednia:

```python
for value in values:
    print(value)
```

Jeżeli indeks nie jest potrzebny, druga wersja jest zwykle prostsza i czytelniejsza.

````python
values = list(range(1000))

%%timeit
for i in range(len(values)):
    x = values[i]
````

````python
%%timeit
for value in values:
    x = value
````

## Co pokazuje benchmark?

Nie chodzi o zapamiętanie liczby mikrosekund.

Interesuje nas:

- pierwszy zapis wykonuje dodatkowe operacje związane z indeksowaniem,
- drugi bezpośrednio pobiera kolejne elementy,
- czytelniejsza konstrukcja może być również prostsza do wykonania.

### Priorytet

**poprawność → czytelność → właściwa konstrukcja → wydajność**.

## `enumerate()`

Często potrzebujemy jednocześnie **elementu i jego numeru**.

Zamiast:

```python
for i in range(len(names)):
    print(i, names[i])
```

używamy:

```python
for i, name in enumerate(names):
    print(i, name)
```

````python
names = ["Anna", "Jan", "Maria"]

for i, name in enumerate(names):
    print(i, name)
````

## Numerowanie od 1

`enumerate()` może rozpocząć numerację od dowolnej wartości.

````python
names = ["Anna", "Jan", "Maria"]

for i, name in enumerate(names, start=1):
    print(i, name)
````

## `zip()`

`zip()` pozwala iterować równolegle po kilku kolekcjach.

````python
names = ["Anna", "Jan", "Maria"]
scores = [5.0, 4.0, 4.5]

for name, score in zip(names, scores):
    print(name, score)
````

## `zip()` i różne długości

Domyślnie `zip()` kończy się wraz z najkrótszym wejściem.

````python
names = ["Anna", "Jan", "Maria"]
scores = [5.0, 4.0]

print(list(zip(names, scores)))
````

## `break`

`break` natychmiast kończy najbliższą pętlę.

````python
numbers = [3, 8, -1, 5]

for number in numbers:
    if number < 0:
        print("Znaleziono wartość ujemną")
        break

    print("Sprawdzam:", number)
````

## `continue`

`continue` pomija resztę bieżącej iteracji i przechodzi do następnej.

````python
numbers = [3, -2, 5, -7, 4]

for number in numbers:
    if number < 0:
        continue

    print(number)
````

## `pass`

`pass` nie wykonuje żadnej operacji.

Przydaje się jako tymczasowe wypełnienie bloku, który składniowo musi istnieć.

````python
condition = True

if condition:
    pass  # kod zostanie dodany później

print("Program działa dalej")
````

## `for ... else`

Python pozwala dodać `else` do pętli.

Blok `else` wykonuje się wtedy, gdy pętla zakończy się **bez użycia `break`**.

````python
values = [2, 4, 6, 8]
target = 5

for value in values:
    if value == target:
        print("Znaleziono")
        break
else:
    print("Nie znaleziono")
````

## Program rośnie razem z naszymi możliwościami

Zaczęliśmy od:

```python
temperature = 31

if temperature > 30:
    print("Alarm")
```

Teraz potrafimy analizować serię pomiarów.

````python
temperatures = [21, 24, 31, 28, 35]
limit = 30

for index, temperature in enumerate(temperatures, start=1):
    if temperature <= limit:
        continue

    print(f"Pomiar {index}: przekroczono limit ({temperature})")
````

## Przykład z wcześniejszym zakończeniem

Możemy przerwać analizę po pierwszym alarmie:

````python
temperatures = [21, 24, 31, 28, 35]
limit = 30

for index, temperature in enumerate(temperatures, start=1):
    if temperature > limit:
        print(f"Pomiar {index}: przekroczono limit")
        break
else:
    print("Wszystkie pomiary poprawne")
````
